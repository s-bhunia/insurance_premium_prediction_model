Exploratory Data Analysis and Visualization

```python
sns.set_style('darkgrid')
matplotlib.rcParams['font.size'] = 14
matplotlib.rcParams['figure.figsize'] = (10, 6)
matplotlib.rcParams['figure.facecolor'] = '#00000000'
```

```python
fig = px.histogram(df, x = 'age', marginal = 'box', nbins = 47, title = 'Distribution og Age') 
fig.update_layout(bargap = 0.1)
fig.show()
```

```python
fig2 = px.histogram(df, x = 'bmi', marginal= 'box', color_discrete_sequence = ['red'], title= "distribution of bmi")
fig2.update_layout(bargap = 0.1)
fig2.show()
```

```python
fig3 = px.histogram(df, x = 'expenses', color = 'smoker', color_discrete_sequence = ['red', 'green'], title = 'annual medical expances')
fig3.update_layout(bargap = 0.1)
fig3.show()
```

```python
fig4 = px.histogram(df, x = 'expenses', color = 'sex', color_discrete_sequence = ['red', 'green'], title = 'annual medical expances')
fig4.update_layout(bargap = 0.1)
fig4.show()
```

```python
fig = px.histogram(df,
                    x="expenses",
                    marginal="box",
                    color="region",
                    color_discrete_sequence=["grey", "orange", "green","violet"],
                    title="Annual Medical Expenses by Region")
fig.update_layout(bargap=0.1)
fig.show()    
```

```python
px.histogram(df, x = 'sex', color = 'smoker', title="Smkoer")
```

```python
px.histogram(df, x = 'region', color = 'smoker', title="smoker vs region")
```

```python
px.histogram(df, x = "children", color = "smoker", title ="smoker vs child" )
```

```python
fig = px.scatter(df,
                x="bmi",
                y="expenses",
                color="smoker",
                hover_data=['sex'],
                opacity=0.8,
                width=800,
                height=500,
                title="BMI Vs Expenses")

fig.update_traces(marker_size=5)
fig.show()   
```

```python
fig = px.scatter(df,
                x="bmi",
                y="expenses",
                color="smoker",
                hover_data=['region'],
                opacity=0.9,
                width=800,
                height=500,
                title= "BMI Vs Expenses")

fig.update_traces(marker_size=5)
fig.show()  
```
